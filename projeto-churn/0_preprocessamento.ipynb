## Preprocessamento de dados - churn

Gera datasets preprocessados com one-hot encoding e RobustScaler.

```python
import os
import pandas as pd
from churn_preprocessing import (load_split_datasets, preprocess_data,
                                 DATA_DIR, INPUT_PREFIX, OUTPUT_PREFIX, TARGET_COL)
```

```python
train_df, val_df, test_df = load_split_datasets(data_dir=DATA_DIR, prefix=INPUT_PREFIX)

(
    X_train, y_train, X_val, y_val, X_test, y_test, feature_names, multi_mappings
) = preprocess_data(train_df, val_df=val_df, test_df=test_df, target_col=TARGET_COL)

feature_names = list(feature_names)
train_processed = pd.DataFrame(X_train, columns=feature_names)
train_processed[TARGET_COL] = y_train
val_processed = pd.DataFrame(X_val, columns=feature_names)
val_processed[TARGET_COL] = y_val
test_processed = pd.DataFrame(X_test, columns=feature_names)
test_processed[TARGET_COL] = y_test


print("Shapes:")
print("Train:", train_processed.shape)
print("Val:", val_processed.shape)
print("Test:", test_processed.shape)
print("Mapeamentos das variáveis categóricas multiclasse:")
for col, mapping in multi_mappings.items():
    print(f"  {col}: {mapping}")
```

```python

train_path = os.path.join(DATA_DIR, f"{OUTPUT_PREFIX}_train.csv")
val_path = os.path.join(DATA_DIR, f"{OUTPUT_PREFIX}_val.csv")
test_path = os.path.join(DATA_DIR, f"{OUTPUT_PREFIX}_test.csv")

train_processed.to_csv(train_path, index=False)
val_processed.to_csv(val_path, index=False)
test_processed.to_csv(test_path, index=False)

print("Datasets preprocessados salvos:")
print(train_path)
print(val_path)
print(test_path)
```
